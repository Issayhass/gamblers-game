```python
import random

#task1
def play_game(start, target, p=0.5):
    number_of_rounds=0
    if start<0:
        return 'Negative values unauthorized'
    else:
        while 0 < start< target:
            if random.random()<p:
                start=start+1
                number_of_rounds=number_of_rounds+1
            else:
                start=start-1
                number_of_rounds=number_of_rounds+1
        final_coins=start       
        return (final_coins, number_of_rounds)
 
#print(play_game(0, 10))

#thinka and explain
jp_m = {}
for i in range(20):
    jp_m[play_game(10, 50, p=0.5)[1]] = i+1
for i in jp_m.keys():
 print(i,end=' ')
longest_game = jp_m[max(jp_m.keys())]
print('\nthe longest game is game number :', longest_game)
shortest_game = jp_m[min(jp_m.keys())]
print('the shortest game is game number :',shortest_game)
```
